# Chapter 6 Return Values
## Some Functions Have Return Values

```python
# When you call a function like math.sqrt, the result is called a return value.
# If the function call appears at the end of a cell, Jupyter displays the return value immediately:
import math  
math.sqrt(42 / math.pi)
```

```python
# If you assign the return value to a variable, it doesn’t get displayed:
radius = math.sqrt(42 / math.pi)
# But you can display it later:
radius
```

```python
radius + math.sqrt(42 / math.pi)
```

```python
def circle_area(radius):  
    area = math.pi * radius**2  
    return area
circle_area(radius)
```

## And Some Have None
- If a function doesn’t have a return statement, it returns None, which is a special value like True and False.

```python
def repeat(word, n):
    print(word * n)
repeat('Filand, ', 3)
# This function uses the print function to display a string, but it does not use a return statement to return a value.
```

```python
result = repeat('Filand, ', 3)
result
# result actually has a value, but Jupyter doesn’t show it. Until we "print" it:
print(result)
```

```python
# here’s a function similar to repeat except that it has a return value:
def repeat_string(word, n):
    return word * n
line = repeat_string('Spam, ', 4)
line
```

- A function like this is called a **pure function** because it doesn’t display anything or have any other effect—other than returning a value.

## Recursion with Return Values


```python
def factorial(n):
    if n == 0:
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        return result
factorial(10)
```

## Fibonacci
- The most common example of a recursive function

```python
def fibonacci(n):
    if n == 0 or n == 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
fibonacci(4)
```

## Checking Types
- We can use the built-in function **isinstance** to check the type of argument.
- input validation

```python
factorial(1.5)
```

```python
isinstance(42, int)
```

```python
isinstance(42, float)
```

```python
def infactorial(n):
    if not isinstance(n, int):
        print('Factorial is only defined for integers.')
        return None
    elif n < 0:
        print('Factorial is not defined for negative integers.')
        return None
    elif n == 0:
        return 1    
    else:
        recurse = infactorial(n-1)
        result = n * recurse
        return result
infactorial(-5.5)
infactorial(-2)
infactorial('crunchy frog')
```

## Debugging


```python
def factorial(n):
    space = ' ' * (4 * n)
    print(space, 'factorial', n)
    if n == 0:
        print(space, 'returning 1')
        return 1
    else:
        recurse = factorial(n-1)
        result = n * recurse
        print(space, 'returning', result)
        return result
factorial(3)
```
