The modulus operator % is more useful than it might seem. 
- it can check whether one number is divisible by another: if x % y is zero, then x is divisible by y. 
- Also, it can extract the rightmost digit or digits from a number. For example, x % 10 yields the rightmost digit of x (in base 10). Similarly, x % 100 yields the last two digits. 
- Finally, the modulus operator can do “clock arithmetic.” For example, if an event starts at 11 A.M. and lasts three hours, we can use the modulus operator to figure out what time it ends:

```python
start = 11
duration = 3
end = (start + duration) % 12
end
# The event would end at 2 P.M.:
```

## If Statements
- One of the Conditional statements
- if statements have the same structure as function definitions: a header followed by an indented statement or sequence of statements called a  block.
- The boolean expression after if is called the condition.

```python
if x > 0:
    print('x is positive')
```

There is no limit to the number of statements that can appear in the block, but there has to be at least one. Occasionally, it is useful to have a block that does nothingusually as a place keeper for code you haven’t written yet. In that case, you can use the  pass statement

```python
if x < 0:
    pass
```

## Recursion
- It is legal for a function to call itself.
- A function that calls itself is recursive.

```python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n-1)
countdown(5)
```

```python
def print_n_times(s, n):
    if n > 0:
        print(s)
        print_n_times(s, n-1)
print_n_times('Spam', 4)
# For simple examples like this, it is probably easier to use a for loop. 
# But we will see examples later that are hard to write with a for loop and easy to write with recursion, so it is good to start early.
```

## Ask a Virtual Assistant
- What is the logical xor operation and how do I compute it in Python?

```python
# The most common and Pythonic way to compute a logical XOR is by using the inequality operator (!=), 
# because two booleans are exclusive if and only if they are not equal:
b = True
a = False

# Logical XOR using inequality
result = (a != b)  # Evaluates to True, or result = a != b
print(result)
```

```python
# Alternatively, you can use operator.xor() from Python's built-in operator module:
import operator

result = operator.xor(True, False)  # Evaluates to True
```

```python
# For Integers (Bitwise XOR)
# If you want to perform an XOR operation on integers (comparing their binary bits), Python uses the caret operator (^):
# Example: 7 XOR 2
result = 7 ^ 2  # Evaluates to 5
print(result)
# For instance, comparing 0111 ($7$) and 0010 ($2$) bit-by-bit results in 0101 ($5$)
```

```python
def recurse(n, s):  
    if n == 0:  
        print(s)  
    else:
        recurse(n-1, n+s)  
recurse(3, 0)
```

```python
from jupyturtle import forward, left, make_turtle, right, back 
make_turtle(delay=0)
def draw(length):  
    angle = 50 
    factor = 0.6  

    if length > 5:  
        forward(length) 
        left(angle) 
        draw(factor * length)  # recursive call!
        right(2 * angle) 
        draw(factor * length) # recursive call!
        left(angle) 
        back(length)
draw(50)
```

```python

def koch(length):
    if length < 5:
        forward(length)
    else:
        koch(length / 3) # recursive call!
        left(60)
        koch(length / 3)
        right(120)
        koch(length / 3)
        left(60)
        koch(length / 3)

make_turtle(delay=0) 
koch(120)
```

```python
from jupyturtle import forward, left, make_turtle


def draw_sierpinski(length, depth):
  half = length / 2
  if depth == 0:
    # Base case: draw a single triangle
    for _ in range(3):
      forward(length)
      left(120)
  else:
    # 1. Draw bottom-left sub-triangle
    draw_sierpinski(half, depth - 1)
    forward(half)

    # 2. Draw bottom-right sub-triangle
    draw_sierpinski(half, depth - 1)
    forward(half)
    left(120)
    forward(half)

    # 3. Draw top sub-triangle
    draw_sierpinski(half, depth - 1)
    forward(half)
    left(120)
    forward(length)
    left(120)


# Initialize the canvas and draw the fractal
make_turtle(delay=0, height=200)
draw_sierpinski(100, 3)
```
