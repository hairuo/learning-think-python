## Encapsulation and Generalization
- Wrapping a piece of code up in a function is called **encapsulation**.

```python
def square():
    for i in range(4):
        forward(50)
        left(90)
make_turtle()
square()
```

- Adding a parameter to a function is called **generalization**

```python
# If we add another parameter, we can make it even more general.
def square(length):
    for i in range(4):
        forward(length)
        left(90)
make_turtle()
square(30)
square(60)
```

```python
# The following function draws regular polygons with a given number of sides:
def polygon(sides, length):
    angle = 360 / sides
    for i in range(sides):
        forward(length)
        left(angle)
make_turtle() 
polygon(7, 30)
```

- keyword arguments (not to be confused with Python keywords like for and def): sides, length

```python
make_turtle() 
polygon(sides=8, length=30) 
```

## Approximating a Circle

```python
import math 
def circle(radius):  
    circumference = 2 * math.pi * radius 
    n = 30 
    length = circumference / n
    polygon(n, length)
make_turtle(delay=0.02) # peed it up by calling make_  turtle with a keyword argument called delay
circle(30)
```

## Refactoring
- In this example, we started with working code and reorganized it with different functions. Changes like this, which improve the code without changing its behavior, are called **refactoring**.

```python
def polyline(n, length, angle):
    for i in range(n):
        forward(length)
        left(angle)

def polygon(n, length):
    angle = 360 / n
    polyline(n, length, angle)

def arc(radius, angle):
    arc_length = 2 * math.pi * radius * abs(angle) / 360
    n = 30
    length = arc_length / n
    step_angle = angle / n
    polyline(n, length, step_angle)  

def circle(radius):
    arc(radius, 360)

make_turtle(delay=0.02) 
polygon(n=20, length=9) 
arc(radius=70, angle=70) 
circle(radius=10)
```

- a few more turtle functions you might want to use:  
- penup  
  - Lift the turtle’s imaginary pen so it doesn’t leave a trail when it moves.  
- pendown  
  - Put the pen back down.

```python
from jupyturtle import penup, pendown
def jump(length):
    """Move forward length units without leaving a trail. 
    Postcondition: Leaves the pen down. 
    """
    penup()
    forward(length)
    pendown()

```

- Write a function called rectangle that draws a rectangle with given side lengths. For example, here’s a rectangle that’s 80 units wide and 40 units tall:

```python
def rectangle(width, height):
    """Draws a rectangle with the given width and height. 
    width:  width of the rectangle 
    height: height of the rectangle 
    """
    for i in range(2):  
        forward(width)
        left(90)
        forward(height)
        left(90)
make_turtle()
rectangle(80, 40)
```

- Write a function called rhombus that draws a rhombus with a given side length and a given interior angle. For example, here’s a rhombus with side length of 50 and an interior angle of 60 degrees:

```python
def rhombus(length, angle):
    """Draws a rhombus with the given side length and interior angle. 
    length:  length of each side 
    angle: interior angle in degrees 
    """
    for i in range(2):  
        forward(length)
        left(angle)
        forward(length)
        left(180 - angle)
make_turtle()
rhombus(50, 60)
```

- write a more general function called parallelogram that draws a quadrilateral with parallel sides. Then rewrite rectangle and rhombus to use parallelogram.

```python
def parallelogram(length1, length2, angle):
    """Draws a parallelogram with the given side lengths and interior angle. 
    length1:  length of the first pair of parallel sides, parallel to the x-axis
    length2:  length of the second pair of parallel sides.
    angle: interior angle in degrees 
    """
    for i in range(2):  
        forward(length1)
        left(angle)
        forward(length2)
        left(180 - angle)
make_turtle(delay=0.02)
jump(-140)
parallelogram(60, 40, 90)
jump(80)
parallelogram(40, 60, 60)
jump(80)
parallelogram(80, 60, 60)
```

- Write an appropriately general set of functions that can draw shapes like this. Hint: write a function called triangle that draws one triangular segment, and then a function called draw_pie that uses triangle.

```python
def triangle(length, angle):
    """Draws a triangle with the given side length and interior angle. 
    length:  length of each side 
    angle: interior angle in degrees 
    """
    forward(length)
    left(90 + angle/2)
    forward(2*length*math.sin(math.radians(angle/2)))
    left(90 + angle/2)
    forward(length)
    left(180)

make_turtle(delay=0.02)
triangle(30, 30)
```

```python
def draw_pie(n, length):
    """Draws a pie with n slices of the given length. 
    n: number of slices 
    length: length of each slice 
    """
    for i in range(n):
        triangle(length, 360/n)
make_turtle(delay=0.02)
jump(-100)
draw_pie(5, 40)
jump(100)
draw_pie(6, 40)
jump(100)
draw_pie(7, 40)
```
