# Chapter 4 Functions and Interfaces
## The jupyturtle Module

```python
# To use the jupyturtle module, we can import it like this:
import jupyturtle
# Now we can use the functions defined in the module, like make_turtle and forward:
jupyturtle.make_turtle()  # make_turtle creates a canvas
jupyturtle.forward(100) # forward moves the turtle a given distance in the direction it’s facing, drawing a line  segment along the way.
```

```python
# We will use functions defined in the jupyturtle module many times, so it would be nice if we did not have to write the name of the module every time. 
# That’s possible if we import the module like this:
from jupyturtle import make_turtle, forward
make_turtle() 
forward(100)
```

## Making a Square

```python
# jupyturtle provides two other functions we’ll use, called left and right. We’ll  import them like this:
from jupyturtle import left, right
make_turtle() 
forward(50) 
left(90) 

forward(50)
right(-90)

forward(50)
left(90)

forward(50)
left(90)
```

```python
make_turtle()
for i in range(4):
    forward(50)
    left(90)
```

## Encapsulation and Generalization
- Wrapping a piece of code up in a function is called **encapsulation**.

```python
def square():
    for i in range(4):
        forward(50)
        left(90)
make_turtle()
square()
```

- Adding a parameter to a function is called **generalization**

```python
def square(length):
    for i in range(4):
        forward(length)
        left(90)
make_turtle()
square(30)
square(60)
```
