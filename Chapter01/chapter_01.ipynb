## Exercises
- You might wonder what round does if a number ends in 0.5. The answer is that it sometimes rounds up and sometimes rounds down.

```python
a = round(42.5); print(a)
b = round(42.5, 1); print(b)
c = round(42.4); print(c)
d = round(42.6); print(d)
e = round(43.5); print(e)
```
